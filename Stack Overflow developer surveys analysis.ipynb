```python
# Завдання 2

# завантаження
df = pd.read_csv("Downloads/stack-overflow-developer-survey-2024/survey_results_public.csv")
schema = pd.read_csv("Downloads/stack-overflow-developer-survey-2024/survey_results_schema.csv")

# список питань (колонок)
questions = schema["qname"]

# залишаємо тільки ті, що є в df
valid_questions = df.columns.intersection(questions)

# респонденти без жодного пропуску
complete_responses = df[valid_questions].notna().all(axis=1)

# кількість таких респондентів
complete_responses.sum()
```

```python
# Завдання 3

import pandas as pd

df = pd.read_csv("Downloads/stack-overflow-developer-survey-2024/survey_results_public.csv")

# перетворення в число (про всяк випадок)
work_exp = pd.to_numeric(df["WorkExp"], errors="coerce")

# обчислення метрик
mean_val = work_exp.mean()
median_val = work_exp.median()
mode_val = work_exp.mode()[0]

# оформлення результату
result = pd.Series({
    "mean": mean_val,
    "median": median_val,
    "mode": mode_val
})

result
```

```python
# Завдання 7

import pandas as pd

df = pd.read_csv("Downloads/stack-overflow-developer-survey-2024/survey_results_public.csv")

# 1. Фільтр: тільки Python-розробники
python_df = df[df["LanguageHaveWorkedWith"].str.contains("Python", na=False)]

# 2. Групування по країнах і агрегація
result = python_df.groupby("Country")["ConvertedCompYearly"].agg(
    mean_comp="mean",
    median_comp="median"
).reset_index()

result
```

```python
# Завдання 8

import pandas as pd

df = pd.read_csv("Downloads/stack-overflow-developer-survey-2024/survey_results_public.csv")

# 1. прибираємо пропуски в компенсації
df_clean = df.dropna(subset=["ConvertedCompYearly"])

# 2. сортуємо за компенсацією (спадання)
top5 = df_clean.sort_values("ConvertedCompYearly", ascending=False).head(5)

# 3. беремо рівень освіти
top5["EdLevel"]
```

```python
# Завдання 9

import pandas as pd

df = pd.read_csv("Downloads/stack-overflow-developer-survey-2024/survey_results_public.csv")

# 1. маска Python-розробників
df["is_python"] = df["LanguageHaveWorkedWith"].str.contains("Python", na=False)

# 2. групування по вікових категоріях
result = df.groupby("Age")["is_python"].mean() * 100

# 3. красивий вигляд
result.sort_values(ascending=False)
```

```python
# Завдання 10

import pandas as pd

df = pd.read_csv("Downloads/stack-overflow-developer-survey-2024/survey_results_public.csv")

# прибираємо NaN у компенсації
comp = df["ConvertedCompYearly"].dropna()

# 75-й перцентиль
p75 = comp.quantile(0.75)

# фільтр
filtered = df[
    (df["ConvertedCompYearly"] > p75) &
    (df["RemoteWork"] == "Remote")
]

# частота індустрій (з очищенням NaN)
top_industries = filtered["Industry"].value_counts(dropna=True)

top_industries
```
